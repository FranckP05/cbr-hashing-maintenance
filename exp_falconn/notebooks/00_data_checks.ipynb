# 00 — Data Checks (exp_falconn)

Sanity checks on the L2-normalised cosine arm for both datasets.


```python
import numpy as np
import pathlib
import sys

root_dir = pathlib.Path().resolve().parent
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))
```

```python
# ============================================================
# CONFIG
# ============================================================
SPLIT_SEED = 42
HEART_CSV_PATH = ''
CHEST_EMBEDDINGS_PATH = ''
CHEST_LABELS_PATH = ''
```

```python
from src.datasets.heart_disease import load_and_split as heart_load
from src.datasets.chest_xray14 import load_and_split as chest_load

H_tr, H_ytr, H_va, H_yva, H_te, H_yte = heart_load(HEART_CSV_PATH, split_seed=SPLIT_SEED, normalise_cosine=True)
C_tr, C_ytr, C_va, C_yva, C_te, C_yte = chest_load(CHEST_EMBEDDINGS_PATH, CHEST_LABELS_PATH, split_seed=SPLIT_SEED, normalise_cosine=True)

for name, X in [('heart-train', H_tr), ('heart-val', H_va), ('heart-test', H_te),
                ('chest-train', C_tr), ('chest-val', C_va), ('chest-test', C_te)]:
    norms = np.linalg.norm(X, axis=1)
    print(f'{name}: shape={X.shape} mean_norm={norms.mean():.4f} std_norm={norms.std():.4f} nan={np.isnan(X).any()} inf={np.isinf(X).any()}')
```

```python
assert np.allclose(np.linalg.norm(H_tr, axis=1), 1.0, atol=1e-4)
assert np.allclose(np.linalg.norm(C_tr, axis=1), 1.0, atol=1e-4)
print('OK: all vectors L2-normalised (cosine arm).')
```
